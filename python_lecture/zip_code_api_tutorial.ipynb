```python
# requestモジュールを使うことを宣言します。
import requests

#
# 郵便番号を入力する処理
#
code = input('郵便番号を入力してください。')
code1 =  code[0:3]
code2 =  code[3:7]

# 郵便番号APIのURLを変数に入れます。
api = 'https://madefor.github.io/postal-code-api/api/v1/{0}/{1}.json'
api = api.format(code1, code2)

# requests.get()で、URLにアクセスして、結果を変数に入れます。
data = requests.get(api)

# 取得したデータを表示します。
print(data.text)
```

```python
jsondata = data.json()
print(jsondata['code'])
```
