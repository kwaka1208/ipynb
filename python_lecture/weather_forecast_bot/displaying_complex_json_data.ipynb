このプログラムで使うJSONデータは[こちら](https://docs.crssrds.jp/assets/data/single.json)でご覧いただけます。

````python
# JSONデータを取り扱うモジュールです
import json

import requests

# 今回は、サイト上にJSONデータをプログラムの中に用意します。
url = 'https://docs.crssrds.jp/assets/data/single.json'

data = requests.get(url)
jsondata = data.json()
print(jsondata)
````

````python
# りんごの情報を表示してみよう

print(jsondata['name'], jsondata['stock'], jsondata['price'])
````

## 複数の果物のデータをもったJSONデータ

このデータは[こちら](https://docs.crssrds.jp/assets/data/multi.json)でご覧いただけます。

```
{
    "fruit":  [
        {"name": "りんご", "stock": 50, "price": 100}, # 0番目
        {"name": "みかん", "stock": 75, "price": 200}, # 1番目
        {"name": "バナナ", "stock": 90, "price": 150}  # 2番目
    ]
}
```

````python
# JSONデータを取得する
url = 'https://docs.crssrds.jp/assets/data/multi.json'

data = requests.get(url)
jsondata = data.json()

# 取得したJSONデータを表示
print(jsondata)
````

````python
# みかんの名前と在庫数、値段を表示してみよう

print(jsondata['fruit'][1]['name'], jsondata['fruit'][1]['stock'], jsondata['fruit'][1]['price'])
````

## 複数のグループに分かれたJSONデータ

このデータは[こちら](https://docs.crssrds.jp/assets/data/multi2.json)でご覧いただけます。


```
{
    "fruit":  [
        {"name": "りんご", "stock": 50, "price": 100},
        {"name": "みかん", "stock": 75, "price": 200},
        {"name": "バナナ", "stock": 90, "price": 150}
    ],
    "vegetable":  [
        {"name": "にんじん", "stock": 20, "price": 300},
        {"name": "はくさい", "stock": 15, "price": 250},
        {"name": "キャベツ", "stock": 10, "price": 320}
    ]
}
```

````python
# JSONデータを取得する
url = 'https://docs.crssrds.jp/assets/data/multi2.json'

data = requests.get(url)
jsondata = data.json()

# 取得したJSONデータを表示
print(jsondata)
````

````python
# 複数の商品からみかんとキャベツの情報を表示してみよう

print(jsondata['fruit'][1]['name'], jsondata['fruit'][1]['stock'], jsondata['fruit'][1]['price'])
print(jsondata['vegetable'][2]['name'], jsondata['vegetable'][2]['stock'], jsondata['vegetable'][2]['price'])
````
