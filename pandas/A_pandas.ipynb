# Pandas 演習: 解答例

列の平均（科目の平均）と行の平均（個人の平均）を計算するプログラムです。

```python
import pandas as pd

# 1. DataFrameを作成します（名前をインデックスにします）
data = {
    "数学": [85, 72, 95],
    "英語": [92, 80, 88]
}
df = pd.DataFrame(data, index=["山本", "鈴木", "高橋"])
print("成績表:")
display(df)

# 2. 数学の平均点を計算します
math_mean = df["数学"].mean()
print(f"\n数学の平均点: {math_mean:.1f} 点")

# 3. 鈴木さんの平均点を計算します
# .loc で行を選択してから平均（mean）を出します
suzuki_mean = df.loc["鈴木"].mean()
print(f"鈴木さんの平均点: {suzuki_mean:.1f} 点")
```
