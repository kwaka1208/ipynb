# じゃんけん勝ち負け判定（二人版）（改）

## 最初にあいこを判定する

```python
import random

# 手の数値を文字列に変換するための辞書（表示用）
hands   = {0: "ぐー", 1: "ちょき", 2: "ぱー"}
player  = random.randint(0, 2) # プレイヤーの手を0〜2でランダムに決定
cp1     = random.randint(0, 2)    # コンピュータの手を0〜2でランダムに決定

print(f"プレイヤーの手: {hands[player]}")
print(f"コンピュータの手: {hands[cp1]}")
print("結果: ", end="")

## プレイヤーとコンピュータの手が同じかどうかを最初に判定する
if (player == cp1):    # プレイヤーとコンピュータの手が同じ場合
    print("あいこ")       # 結果：あいこ

## 残りのパターンをプレイヤーの手ごとに判定する
# プレイヤーがぐー
if (player == 0): 
    if (cp1 == 1):     # 2 プレイヤー：ぐー   コンピュータ：ちょき
        print("勝った")   # 結果：勝った
    elif (cp1 == 2):     # 3 プレイヤー：ぐー   コンピュータ：ぱー
        print("負けた")   # 結果：負けた

# プレイヤーがちょき
if (player == 1): 
    if (cp1 == 0):       # 4 プレイヤー：ちょき  コンピュータ：ぐー
        print("負けた")   # 結果：負けた
    elif (cp1 == 2):     # 6 プレイヤー：ちょき  コンピュータ：ぱー
        print("勝った")   # 結果：勝った

# プレイヤーがぱー
if (player == 2): 
    if (cp1 == 0):       # 7 プレイヤー：ぱー   コンピュータ：ぐー
        print("勝った")   # 結果：勝った
    elif (cp1 == 1):     # 8 プレイヤー：ぱー   コンピュータ：ちょき
        print("負けた")   # 結果：負けた
```

## あいこを判定した後、勝った時だけを判定する。それ以外は負けた

```python
import random

# 手の数値を文字列に変換するための辞書（表示用）
hands   = {0: "ぐー", 1: "ちょき", 2: "ぱー"}
player  = random.randint(0, 2) # プレイヤーの手を0〜2でランダムに決定
cp1     = random.randint(0, 2)    # コンピュータの手を0〜2でランダムに決定

print(f"プレイヤーの手: {hands[player]}")
print(f"コンピュータの手: {hands[cp1]}")
print("結果: ", end="")

if (player == cp1):    # プレイヤーとコンピュータの手が同じ場合
    print("あいこ")       # 結果：あいこ
elif (player == 0 and cp1 == 1) or (player == 1 and cp1 == 2) or (player == 2 and cp1 == 0):
    print("勝った")       # 結果：勝った
else:
    print("負けた")       # 結果：負けた
```
