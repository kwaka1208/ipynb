# じゃんけん勝ち負け判定（二人版）



```python
import random

# 手の数値を文字列に変換するための辞書（表示用）
hands   = {0: "ぐー", 1: "ちょき", 2: "ぱー"}
player  = random.randint(0, 2) # プレイヤーの手を0〜2でランダムに決定
cp1     = random.randint(0, 2)    # コンピュータの手を0〜2でランダムに決定

print(f"プレイヤーの手: {hands[player]}")
print(f"コンピュータの手: {hands[cp1]}")
print("結果: ", end="")

# プレイヤーがぐー
if (player == 0): 
    if (cp1 == 0):       # 1 プレイヤー：ぐー   コンピュータ：ぐー
        print("あいこ")   # 結果：あいこ
    elif (cp1 == 1):     # 2 プレイヤー：ぐー   コンピュータ：ちょき
        print("勝った")   # 結果：勝った
    elif (cp1 == 2):     # 3 プレイヤー：ぐー   コンピュータ：ぱー
        print("負けた")   # 結果：負けた

# プレイヤーがちょき
if (player == 1): 
    if (cp1 == 0):       # 4 プレイヤー：ちょき  コンピュータ：ぐー
        print("負けた")   # 結果：負けた
    elif (cp1 == 1):     # 5 プレイヤー：ちょき  コンピュータ：ちょき
        print("あいこ")   # 結果：あいこ
    elif (cp1 == 2):     # 6 プレイヤー：ちょき  コンピュータ：ぱー
        print("勝った")   # 結果：勝った

# プレイヤーがぱー
# if (player == 2): 
    if (cp1 == 0):       # 7 プレイヤー：ぱー   コンピュータ：ぐー
        print("勝った")   # 結果：勝った
    elif (cp1 == 1):     # 8 プレイヤー：ぱー   コンピュータ：ちょき
        print("負けた")   # 結果：負けた
    elif (cp1 == 2):     # 9 プレイヤー：ぱー   コンピュータ：ぱー
        print("あいこ")   # 結果：あいこ
```
