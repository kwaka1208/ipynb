# じゃんけん勝ち負け判定（三人版）


```python
import random

# 手の数値を文字列に変換するための辞書（表示用）
hands   = {0: "ぐー", 1: "ちょき", 2: "ぱー"}
player  = random.randint(0, 2) # プレイヤーの手を0〜2でランダムに決定
cp1     = random.randint(0, 2) # コンピュータ1の手を0〜2でランダムに決定
cp2     = random.randint(0, 2) # コンピュータ2の手を0〜2でランダムに決定

print(f"プレイヤーの手: {hands[player]}")
print(f"コンピュータ1の手: {hands[cp1]}")
print(f"コンピュータ2の手: {hands[cp2]}")
print("結果: ", end="")

# プレイヤーがぐー
if (player == 0): 
    if (cp1 == 0):
        if (cp2 == 0):       # 1 プレイヤー：ぐー   cp1：ぐー   cp2：ぐー
            print("あいこ")   
        elif (cp2 == 1):     # 2 プレイヤー：ぐー   cp1：ぐー   cp2：ちょき
            print("勝った")   
        elif (cp2 == 2):     # 3 プレイヤー：ぐー   cp1：ぐー   cp2：ぱー
            print("負けた")   
    elif (cp1 == 1):
        if (cp2 == 0):       # 4 プレイヤー：ぐー   cp1：ちょき cp2：ぐー
            print("勝った")   
        elif (cp2 == 1):     # 5 プレイヤー：ぐー   cp1：ちょき cp2：ちょき
            print("勝った")   
        elif (cp2 == 2):     # 6 プレイヤー：ぐー   cp1：ちょき cp2：ぱー
            print("あいこ")   
    elif (cp1 == 2):
        if (cp2 == 0):       # 7 プレイヤー：ぐー   cp1：ぱー   cp2：ぐー
            print("負けた")   
        elif (cp2 == 1):     # 8 プレイヤー：ぐー   cp1：ぱー   cp2：ちょき
            print("あいこ")   
        elif (cp2 == 2):     # 9 プレイヤー：ぐー   cp1：ぱー   cp2：ぱー
            print("負けた")   

# プレイヤーがちょき
if (player == 1): 
    if (cp1 == 0):
        if (cp2 == 0):       # 10 プレイヤー：ちょき cp1：ぐー   cp2：ぐー
            print("負けた")   
        elif (cp2 == 1):     # 11 プレイヤー：ちょき cp1：ぐー   cp2：ちょき
            print("負けた")   
        elif (cp2 == 2):     # 12 プレイヤー：ちょき cp1：ぐー   cp2：ぱー
            print("あいこ")   
    elif (cp1 == 1):
        if (cp2 == 0):       # 13 プレイヤー：ちょき cp1：ちょき cp2：ぐー
            print("負けた")   
        elif (cp2 == 1):     # 14 プレイヤー：ちょき cp1：ちょき cp2：ちょき
            print("あいこ")   
        elif (cp2 == 2):     # 15 プレイヤー：ちょき cp1：ちょき cp2：ぱー
            print("勝った")   
    elif (cp1 == 2):
        if (cp2 == 0):       # 16 プレイヤー：ちょき cp1：ぱー   cp2：ぐー
            print("あいこ")   
        elif (cp2 == 1):     # 17 プレイヤー：ちょき cp1：ぱー   cp2：ちょき
            print("勝った")   
        elif (cp2 == 2):     # 18 プレイヤー：ちょき cp1：ぱー   cp2：ぱー
            print("勝った")   

# プレイヤーがぱー
if (player == 2): 
    if (cp1 == 0):
        if (cp2 == 0):       # 19 プレイヤー：ぱー   cp1：ぐー   cp2：ぐー
            print("勝った")   
        elif (cp2 == 1):     # 20 プレイヤー：ぱー   cp1：ぐー   cp2：ちょき
            print("あいこ")   
        elif (cp2 == 2):     # 21 プレイヤー：ぱー   cp1：ぐー   cp2：ぱー
            print("勝った")   
    elif (cp1 == 1):
        if (cp2 == 0):       # 22 プレイヤー：ぱー   cp1：ちょき cp2：ぐー
            print("あいこ")   
        elif (cp2 == 1):     # 23 プレイヤー：ぱー   cp1：ちょき cp2：ちょき
            print("負けた")   
        elif (cp2 == 2):     # 24 プレイヤー：ぱー   cp1：ちょき cp2：ぱー
            print("負けた")   
    elif (cp1 == 2):
        if (cp2 == 0):       # 25 プレイヤー：ぱー   cp1：ぱー   cp2：ぐー
            print("勝った")   
        elif (cp2 == 1):     # 26 プレイヤー：ぱー   cp1：ぱー   cp2：ちょき
            print("負けた")   
        elif (cp2 == 2):     # 27 プレイヤー：ぱー   cp1：ぱー   cp2：ぱー
            print("あいこ")
```
