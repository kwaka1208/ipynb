# じゃんけん勝ち負け判定（多人数版）


```python
import random

# 手の数値を文字列に変換するための辞書（表示用）
hands   = {0: "ぐー", 1: "ちょき", 2: "ぱー"}
MAX_COMPUTER = 10
cp = []
player  = random.randint(0, 2) # プレイヤーの手を0〜2でランダムに決定
print(f"プレイヤーの手: {hands[player]}")
for i in range(MAX_COMPUTER):
    cp.append(random.randint(0, 2)) # コンピュータの手を0〜2でランダムに決定
    print(f"コンピュータ{i+1}の手: {hands[cp[i]]}")
print("結果: ", end="")

hands_list = [0,0,0]    # ぐー、ちょき、ぱーのリストを作成する
hands_list[player] = 1 # プレイヤーの手の場所に1をセット
for i in range(MAX_COMPUTER):
    hands_list[cp[i]] = 1 # コンピュータの手の場所に1をセット

result = sum(hands_list)   # 出た手の数を合計する
if (result == 3) or (result == 1): # ぐー、ちょき、ぱーのいずれかが3つ揃っている場合、または1つしか出ていない場合
    print("あいこ")
elif (player == 0) and (hands_list[1] == 1): # プレイヤーがぐーで、他にちょきが出ている場合
    print("勝った")
elif (player == 1) and (hands_list[2] == 1): # プレイヤーがちょきで、他にぱーが出ている場合
    print("勝った")
elif (player == 2) and (hands_list[0] == 1): # プレイヤーがぱーで、他にぐーが出ている場合
    print("勝った")
else:
    print("負けた")
```
