```python
import random

count = 0
while True:
    count += 1
    # 1〜100のランダムな数字を生成
    result = random.randint(1, 100)
    if result == 1: # 1%の確率
        print(f"SSRが出ました！かかった回数: {count}回")
        print(f"消費した金額: {count * 300}円")
        break
```

```python
import random

count = 0
result = 0
while result != 1:
    count += 1
    # 1〜100のランダムな数字を生成
    result = random.randint(1, 100)

print(f"SSRが出ました！かかった回数: {count}回")
print(f"消費した金額: {count * 300}円")
```
