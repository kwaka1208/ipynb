```python
import random

base_data = []
data = []
max_data = 10000

for i in range(max_data):
  base_data.append(i)

while len(base_data) != 0:
  index = random.randrange(len(base_data))
  data.append(base_data[index])
  # print(index,base_data[index])
  del base_data[index]

print(len(base_data))
print(data)
```

## 線形探索

```python
data = [1,2,3,4,5,6,7,8]

target = 8
for i in range(len(data)):
  if data[i] == target:
    print("found at",i)
    break
```
