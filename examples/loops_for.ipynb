````python
#
# リストの中の分だけ繰り返す
#
month = ["睦月","如月","弥生","卯月","皐月","水無月","文月","葉月","長月","神無月","霜月","師走"]

for m in month:
  print(m)
````

````python
#
# 決まった数だけ繰り返す
#
for i in range(10): # 0から始まる10個の数字
  print(i)
````

````python
#
# 指定した範囲の数だけ繰り返す
#
for i in range(1,10): # 1〜9まで繰り返す（10ではないところに注意）
    print(i)
````

````python
#
# 二つの繰り返しを組み合わせる
#

for i in range(1,10):
  print("外のループ：", i)
  for j in range(1, 10):
    print("　内のループ：",j)
````

````python
#
# 九九の一の段
#
for i in range(1,10): # 1〜9まで繰り返す（10ではないところに注意）
    print("1 *", i,"=", 1*i)
````

## 作ってみよう

ここまでのプログラムを参考に九九の一の段から九の段までを表示するプログラムを作ってみよう。


````python
#
# 一の段を表示するプログラムを改造して、
# 九九の一の段から九の段までを表示してみよう
# 
for i in range(1,10): # 1〜9まで繰り返す（10ではないところに注意）
    print("1 *", i,"=", 1*i)
````

回答例：九九の一の段から九の段までを表示してみよう
```
for i in range(1,10):
  for j in range(1,10):
    print ( i, "*", j, "=", i*j)
```

## 参考）九九を表形式で表示してみよう

こちらは、九九を表形式で表示するプログラムです。
通常print命令で表示を行うと改行が行われますが、print命令のカッコの中にend オプションを指定すると改行させないこともできます。

たとえば、 print(i*j, end= " ") であれば、改行の代わりに半角スペースで区切られます。
逆に、改行だけを行いたい場合は'\n'を用います。'\'はバックスラッシュと呼ばれる記号でWindowsなどでは半角の￥が表示される場合もあります。

```
for i in range(1,10):
  for j in range(1,10):
    print(i*j, end= " ")
  print('\n')
```
