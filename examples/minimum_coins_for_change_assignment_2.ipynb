```python
price = 872
max1 = price // 1
print('1円玉の最大枚数=', max1,'枚')
max10 = price // 10
print('10円玉の最大枚数=', max10,'枚')
max50 = price // 50
print('50円玉の最大枚数=', max50,'枚')
max100 = price // 100
print('100円玉の最大枚数=', max100,'枚')
max400 = price // 400
print('400円玉の最大枚数=', max400,'枚')
max500 = price // 500
print('500円玉の最大枚数=', max500,'枚')

coins = [0,0,0,0,0,0]
coin_num = float('inf')
count = 0

for num1 in range(0, max1 + 1):
  for num10 in range(0, max10 + 1):
    for num50 in range(0, max50 + 1):
      for num100 in range(0, max100 + 1):
        for num400 in range(0, max400 + 1):
          for num500 in range(0, max500 + 1):
            count = count + 1
            if (num1 + num10 * 10 + num50 * 50  + num100 * 100 + num400 * 400 + num500 * 500)  == price:
              num = num1 + num10 + num50 + num100 + num400 + num500
              if coin_num > num:
                coin_num = num
                coins[0] = num1
                coins[1] = num10
                coins[2] = num50
                coins[3] = num100
                coins[4] = num400
                coins[5] = num500

print('一番少ない硬貨の枚数は')
print('1円玉', coins[0], '枚')
print('10円玉', coins[1], '枚')
print('50円玉', coins[2], '枚')
print('100円玉', coins[3], '枚')
print('400円玉', coins[4], '枚')
print('500円玉', coins[5], '枚')
print('合計',coin_num, '枚')
print('処理回数は','{:,}'.format(count), '回でした')
```
