```python
#
# importの使い方
#
#
# randomモジュールを元の名前のまま使う
# これが一番一般的な使い方
#
import random
print(random.randrange(1,6))
```

```python
#
# randomモジュールを'rnd'という名前で使う
# モジュールの名前が長い時にモジュール名を省略したい時に使う
#
import random as rnd
print(rnd.randrange(1,6))
```

```python
#
# randomモジュールからrandrange関数だけを使う
# 必要な関数が決まっていればこれでも良い。
#
from random import randrange
print(randrange(1,6))
```

## randomを使ったプログラム例

```python
#
# Python サイコロ
#
import random

dice = random.randrange(1,6)
print(dice)
```

```python
#
# Pythonおみくじ
#
import random

omikuji = random.choice(["大吉","中吉","小吉","吉","凶"])
print(omikuji)
```

```python
#
# 俳句メーカー
#
from random import choice as erabu

kami = ['古池や','柿くえば', '閑(しず)かさや','すずめの子']
naka = ['蛙（かわず）とびこむ','鐘が鳴るなり','岩にしみ入る','そこのけそこのけ']
shimo = ['水の音','法隆寺','蝉の声','お馬が通る']

print (erabu(kami), erabu(naka), erabu(shimo) )
```
