2つの数字を入力して、その積（かけざん）を表示するプログラムを作ってみましょう。

- 変数名は自由につけてください。aとかbとかで構いません。
- 入力された数字を整数に変えることを忘れずに。


```python
a = input('かけられる数は：')
a = int(a)
b = input('かける数は：')
b = int(b)
print('a x b = ', a * b)
```
