# Pythonで暗号化を体験しよう
暗号化と復号化（暗号化された情報を元に戻すこと）を実際に動くプログラムで体験してみましょう。

## 鍵を作ろう
まず最初に暗号を作るための鍵を作ります。ここでは、プログラムで一発に鍵が作れるようになっています。


```python
#
# 鍵を作るプログラム
#
from cryptography.fernet import Fernet

key = Fernet.generate_key()
print('暗号化するための鍵を作りました')
print(key.decode())
print('です')
fernet = Fernet(key)
```

## 入力したデータを暗号化しよう
作った鍵で自分の言葉や文章を暗号化してみます

```python
data = input('暗号化したい言葉や文章を入力してください')
crypt_text = fernet.encrypt(data.encode())
print("あなたが入力した文章は以下の通り暗号化されました")
print(crypt_text.decode())
```

## 暗号を解読しよう
作った暗号を解読します。解読するには同じ鍵が必要なので、鍵が変わってしまうと解読できなくなるので要注意です。

```python
data = input('暗号化された文章を入力してください')
try :
  text = fernet.decrypt(data).decode()
  print(f"あなたが入力した文章は「{text}」ですね")
except Exception as e:
  print('入力された暗号は元に戻せませんでした')
  print(e)
```
