```python
!pip install -q openai-whisper
import whisper
from google.colab import drive
import os
from pathlib import Path

def get_audio_files(folder_path):
    return [f for f in os.listdir(folder_path) if f.lower().endswith(('.wav', '.mp3', '.m4a', '.flac', '.ogg'))]

def transcribe_audio(file_path):
    model = whisper.load_model("base")
    result = model.transcribe(file_path)
    return result['text']

def save_transcription(text, output_path):
    with open(output_path, "w", encoding="utf-8") as file:
        file.write(text)

def process_audio_files(audio_folder):
    audio_files = get_audio_files(audio_folder)
    for file_name in audio_files:
        audio_path = os.path.join(audio_folder, file_name)
        transcription_text = transcribe_audio(audio_path)
        output_text_path = os.path.join(audio_folder, Path(file_name).stem + ".txt")
        save_transcription(transcription_text, output_text_path)
        print(f"文字起こし完了: {output_text_path}")

# Google Driveをマウントしてwhisperフォルダを指定
drive.mount('/content/drive')
audio_folder = "/content/drive/MyDrive/whisper"
process_audio_files(audio_folder)
```
