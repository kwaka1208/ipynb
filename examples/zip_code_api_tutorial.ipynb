```python
# requestモジュールを使うことを宣言します。
import requests

#
# 郵便番号を入力する処理
#
code = input('郵便番号を入力してください。')
code1 =  code[0:3]
code2 =  code[3:7]

# 郵便番号APIのURLを変数に入れます。
api = 'https://madefor.github.io/postal-code-api/api/v1/{0}/{1}.json'
api = api.format(code1, code2)

# requests.get()で、URLにアクセスして、結果を変数に入れます。
data = requests.get(api)

# 取得したデータを表示します。
print(data.text)
```

```python
# 全角と半角を変換するライブラリ"zenhan"をインストールします。
!pip install zenhan

# requestモジュールを使うことを宣言します。
import requests

# ライブラリ"zenhan"を使用することを予め記述しておきます。
import zenhan

# 正規表現モジュールを使うための宣言を行います。
import re

#
# 郵便番号を入力する処理
#
code = input('郵便番号を入力してください。')
# チェック用のパターンをzip_patternという変数に作成します。
zip_pattern = re.compile(r"^([0-9]{3})-?([0-9]{4})$")

# 作成したパターンで変数zipの郵便番号をチェックします。
match = zip_pattern.search(code)

# チェック結果を表示します。
print(match)

# 一致しなかった場合は、エラーメッセージを表示します。
if match == None:
  print("入力された郵便番号が正しくありません。もう一度最初からお願いします")
else:
  # 郵便番号APIのURLを変数に入れます。
  api = 'https://madefor.github.io/postal-code-api/api/v1/{0}/{1}.json'
  api = api.format(match.group(1), match.group(2))
	# requests.get()で、URLにアクセスして、結果を変数に入れます。
  data = requests.get(api)
  if data.status_code == 200:
	  # 取得したデータを表示します。
	  print(data.text)
  else:
	  print('住所情報が取得できませんでした、郵便番号が間違っているかもしれません。')
```
