## 課題
乱数を使って1〜100の数字を発生させ、出た数字を表示してください。
そして、その数字が偶数なら「偶数です」奇数なら「奇数です」と表示するプログラムを作ってください。

```
例）
10 偶数です
23 奇数です
```

- 参考になるページ
    - [01 - 文字と数字の表示と演算子](https://docs.crssrds.jp/learn/python/01/)
    - [04 - モジュールの使い方](https://docs.crssrds.jp/learn/python/04/)


````python
#
# 解答例（条件分岐で処理した場合）
#
import random

number = random.randrange(1,101) # 1〜100までの乱数を作る
odd_or_even = number % 2

if odd_or_even == 0:
  print(number, '偶数です')
else:
  print(number, '奇数です')
````

````python
#
# 解答例（条件分岐を使わずにリストで処理した場合）
#
import random

number = random.randrange(1,101) # 1〜100までの乱数を作る
odd_or_even = number % 2

kind = ['偶数', '奇数']

print(number, kind[odd_or_even], 'です')
````
