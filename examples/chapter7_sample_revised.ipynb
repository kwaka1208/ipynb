```python
# バブルソートを行う関数
def bubble_sort(arr):
  count = 0
  n = len(arr)          # nにリストの長さを代入
  for i in range(n-1):            # 外側ループ: 巡はn-1回まで
    for j in range(n-1-i):      # 内側ループ: 比較と交換
      if arr[j] > arr[j+1]:              # 隣り合う要素を比較し，# 順序が正しくない場合は交換
        arr[j], arr[j+1] = arr[j+1], arr[j]
      count = count + 1
      print(f"{i+1}巡目のソート- { j+1 }: {arr}")      # 途中経過を表示
      # time.sleep(wait)
    print("--------------")
  print("終了しました")
  print("処理回数",count)


# リストの長さを設定
LIST_LENGTH = 10

# リストを生成して表示
number_list = [29, 73, 68, 72, 10, 3, 70, 98, 65, 44]
# number_list =  [8, 14, 28, 46, 73, 79, 81, 90, 93, 97]
# number_list = make_list(LIST_LENGTH)
print(f"ソートするリスト: {number_list}")
print("--------------")

# バブルソート関数を呼び出してソートを実行
bubble_sort(number_list)

# ソート後のリストを表示
print(f"ソート後のリスト: {number_list}")
```

## 選択ソート


```python
# 選択ソートを行う関数
def selection_sort(arr):
  count = 0
  n = len(arr)
  for i in range(n-1):              # 外側ループ: 巡はn-1回まで
    min_index = i                     # 注目要素を仮の最小値とする
    for j in range(i+1, n):         # 内側ループ: 注目要素の後ろの要素から最小値を探す。
      print(f"最小値: {min_index}")      # 途中経過を表示
      if arr[j] < arr[min_index]:            # 現在の最小値よりも小さい場合は
        min_index = j                        # 最小値の位置を更新
      print(f"{i+1}巡目のソート- { j+1 }: {arr}")      # 途中経過を表示
      count = count + 1
      # time.sleep(wait)
    arr[i], arr[min_index] = arr[min_index], arr[i]  # 最小値を注目要素と交換
    print("--------------")
  print("終了しました")
  print("処理回数",count)

# リストの長さを設定
LIST_LENGTH = 10

# リストを生成して表示
number_list = [29, 73, 68, 72, 10, 3, 70, 98, 65, 44]
# number_list =  [8, 14, 28, 46, 73, 79, 81, 90, 93, 97]
# number_list = make_list(LIST_LENGTH)
print(f"ソートするリスト: {number_list}")
print("--------------")

# 選択ソート関数を呼び出してソートを実行
selection_sort(number_list)

# ソート後のリストを表示
print(f"ソート後のリスト: {number_list}")
```

挿入ソート

```python
# 挿入ソートを行う関数
def insertion_sort(arr):
  count = 0
  n = len(arr)
  for i in range(1, n):                     # 外側ループ：2番目〜最後の要素　iは注目要素のインデックス
    key = arr[i]                            # 注目要素の値（挿入する値）を変数keyに一時的に保存
    j = i - 1                               # 注目要素の１つ前の要素が比較対象　jは比較対象のインデックス
    while j >= 0 and arr[j] > key:
      arr[j+1] = arr[j]                   # 比較対象の値を1つ後ろにずらす
      j -= 1                                # 比較対象のインデックスを１つ前にずらす
      count = count + 1
      print(f"{i+1}巡目のソート- { j+1 }: {arr}")      # 途中経過を表示
      # time.sleep(wait)
    arr[j + 1] = key                        # 値を適切な位置に挿入
    print("--------------")
  print("終了しました")
  print("処理回数",count)

# リストの長さを設定
LIST_LENGTH = 10

# リストを生成して表示
number_list = [29, 73, 68, 72, 10, 3, 70, 98, 65, 44]
# number_list =  [8, 14, 28, 46, 73, 79, 81, 90, 93, 97]
# number_list = make_list(LIST_LENGTH)
print(f"ソートするリスト: {number_list}")
print("--------------")

# 挿入ソート関数を呼び出してソートを実行
insertion_sort(number_list)

# ソート後のリストを表示
print(f"ソート後のリスト: {number_list}")
```
