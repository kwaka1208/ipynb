````python
#
# 決まった数だけ繰り返す(for)
#
for i in range(10): # 0から始まる10個の数字
  print(i)
````

````python
#
# 決まった数だけ繰り返す(while)
#
# for i in range(10): # 0から始まる10個の数字
#  print(i)

i  = 0
while i < 10:
  print(i)
  i = i + 1
````

````python
#
# 指定した範囲の数だけ繰り返す(for)
#
for i in range(1,10): # 1〜9まで繰り返す（10ではないところに注意）
    print(i)
````

````python
#
# 指定した範囲の数だけ繰り返す(while)
#
i  = 1
while i < 10:
  print(i)
  i = i + 1
````

````python
#
# 正しいパスワードを入れた時だけ終了するプログラム
#

# password:入力されたパスワード
# 最初は空にしておく
password = ""

# パスワードが一致しなければ繰り返す
while password != "pass2023":
  password = input("パスワードを入力してください")
print("----")
print("OK：パスワードを確認しました")
print("処理を終了します")
````

### 作ってみよう解答例

```
#
# Pythonおみくじ
#
import random

omikuji = ["大吉","中吉","小吉","吉","凶"]
hiku = ""

while hiku != "大吉":
  print("おみくじを引きます")
  hiku = random.choice(omikuji)
  print(hiku)
```
