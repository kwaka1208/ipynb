## ヒント1: 改行しない print
print関数は、通常実行ごとに改行が入りますが、`end=""` をつけると改行しないようにできます。

````python
print("通常はprintごとに改行入ります")
print("A")
print("B")
print("C")

print('end=""をつけると改行なしにできます。 ')
print("A", end="")
print("B", end="")
print("C", end="")
````

## ヒント2: 改行コード "\n" の利用
文字列の中で `\n` を使うと、そこで改行させることができます。一つの `print` 関数で複数行を表示したり、空行を入れたりするのに便利です。

````python
print("1行目\n2行目")
print("\n先頭にいれると、空行が入ります")
````

## ヒント3: 数値の桁を揃える
数字をきれいに並べるには、表示する文字の幅（桁数）を揃えると便利です。  
f-string（フォーマット済み文字列）を使うと、`f"{変数名:幅}"` のようにして、指定した文字数分の幅で右寄せ表示ができます。

````python
num1 = 5
num2 = 81

print("幅を指定しない場合:")
print(f"{num1}")
print(f"{num2}")

print("\n幅を3に指定した場合:")
print(f"{num1:3}")
print(f"{num2:3}")
````

## ヒント4: pandasをモジュールの機能を使う
Pythonのデータ分析ライブラリ `pandas` を使うと、2次元のデータ（リストのリストなど）を簡単に表形式（DataFrame）にできます。  
行の見出し（index）や列の見出し（columns）を自由に設定でき、ノートブック上では非常にきれいに表示されます。

````python
import pandas as pd

# 3x3の小さな九九の表データの例
data = [
    [1, 2, 3],
    [2, 4, 6],
    [3, 6, 9]
]

# DataFrameを作成。indexやcolumnsに見出しを指定できます。
df = pd.DataFrame(data, index=[1, 2, 3], columns=[1, 2, 3])

print("DataFrameを表示します:")
display(df)
````
