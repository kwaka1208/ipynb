# 応用演習：九九の表を作ろう

## プログラム例（１）

````python
# 横の見出し（列番号）を表示
print("   1   2   3   4   5   6   7   8   9")

# 1〜9の段をループ
for i in range(1, 10):
    # 縦の見出し（行番号）を表示
    print(i, end="")
    
    # 1列目（×1）の計算結果のみ、幅を3文字分取って表示
    print(f"{i * 1:3}", end="")
    
    # 2〜9列目（×2〜×9）の計算結果は、幅を4文字分取って表示
    for j in range(2, 10):
        print(f"{i * j:4}", end="")
        
    # 1段終わるごとに改行
    print()
````

## プログラム例（2）
numpy と pandasを使って九九の表を作ることもできます。以下はその例です。

```python

````python
import numpy as np
import pandas as pd

# 1. 計算のための「数値の列」と「数値の行」を準備します
# np.arange(1, 10) は [1, 2, ..., 9] という数値の並びを作ります
# .reshape(-1, 1) をつけると、横並びのデータを「縦1列」の形に変換できます
row = np.arange(1, 10).reshape(-1, 1) # 縦 (9行1列)
col = np.arange(1, 10)                # 横 (1行9列)

# 2. 「縦」と「横」を掛け合わせます
# numpyの「ブロードキャスト」という機能により、自動的に9x9の掛け算表（マトリックス）が作られます
matrix = row * col

# 3. 作成したデータをPandasの表（DataFrame）に変換します
# matrix: 元となる数値データ
# index: 行の見出し（左端の1〜9）
# columns: 列の見出し（上端の1〜9）
df = pd.DataFrame(matrix, index=range(1, 10), columns=range(1, 10))

# 4. 結果を表示します
print(df)
````
