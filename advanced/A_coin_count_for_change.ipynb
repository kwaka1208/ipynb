# お釣りの枚数を計算する

あなたはコンビニエンスストアで利用するレジの開発者です。

おつりを自動的に出す機能を担当しています、購入金額と支払い金額からおつりの金額と必要な硬貨の枚数を計算するプログラムを作ってください。

硬貨の枚数は十分あるものとして、硬貨の枚数は一番少なくなるようにします。

例）おつりが8円だった場合
- OK：5円玉　1枚、1円玉 3枚
- NG：1円玉 8枚


## 解答例1：繰り返しを使わない方法

```python
# 硬貨の種類
coin_kind = [500, 100, 50, 10, 5, 1]
# 支払い金額
payment = 1000

# input()で受け取った値は文字列になるため、int()で整数に変換します
amount = int(input('買い物金額を1,000円以下で入力してください: '))

# おつり
change = payment - amount

# 変数を埋め込む場合は f文字列（f"..."）を使用します
print(f'おつりは、{change}円です。')

# 各硬貨の枚数を保存するリストを初期化します（例: [0, 0, 0, 0, 0, 0]）
coin_count = [0] * len(coin_kind)

# for文を使わずに、上から順番に計算します
# 割り算の商（//）で枚数を求め、余り（%）を残りの金額として更新します

# 500円玉の計算
coin_count[0] = change // 500
change = change % 500

# 100円玉の計算
coin_count[1] = change // 100
change = change % 100

# 50円玉の計算
coin_count[2] = change // 50
change = change % 50

# 10円玉の計算
coin_count[3] = change // 10
change = change % 10

# 5円玉の計算
coin_count[4] = change // 5
change = change % 5

# 1円玉の計算
coin_count[5] = change // 1
change = change % 1

# 結果を表示します
print("お釣りで渡す硬貨の枚数は次のとおりです。")
print(f"500円玉の枚数 = {coin_count[0]} 枚")
print(f"100円玉の枚数 = {coin_count[1]} 枚")
print(f" 50円玉の枚数 = {coin_count[2]} 枚")
print(f" 10円玉の枚数 = {coin_count[3]} 枚")
print(f"  5円玉の枚数 = {coin_count[4]} 枚")
print(f"  1円玉の枚数 = {coin_count[5]} 枚")
```

## 解答例2：繰り返し処理で行う方法

```python
# 硬貨の種類
coin_kind = [500, 100, 50, 10, 5, 1]
# 支払い金額
payment = 1000

# input()で受け取った値は文字列になるため、int()で整数に変換します
amount = int(input('買い物金額を1,000円以下で入力してください: '))

# おつり
change = payment - amount

# 変数を埋め込む場合は f文字列（f"..."）を使用します
print(f'おつりは、{change}円です。')

# 各硬貨の枚数を保存するリストを初期化します（例: [0, 0, 0, 0, 0, 0]）
coin_count = [0] * len(coin_kind)

# お釣りの各硬貨の枚数を計算する
for i in range(len(coin_kind)):
    # 割り算の商（//）で硬貨の枚数を求めます
    coin_count[i] = change // coin_kind[i]
    # 割り算の余り（%）を次の硬貨の計算に回します
    change = change % coin_kind[i]

# 結果を表示します
print("お釣りで渡す硬貨の枚数は次のとおりです。")
print(f"500円玉の枚数 = {coin_count[0]} 枚")
print(f"100円玉の枚数 = {coin_count[1]} 枚")
print(f" 50円玉の枚数 = {coin_count[2]} 枚")
print(f" 10円玉の枚数 = {coin_count[3]} 枚")
print(f"  5円玉の枚数 = {coin_count[4]} 枚")
print(f"  1円玉の枚数 = {coin_count[5]} 枚")
```
