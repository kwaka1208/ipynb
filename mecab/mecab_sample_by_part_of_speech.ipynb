```
import MeCab

def mecab_json(text):
  mecab = MeCab.Tagger()
  # テキストをパース
  parsed_lines = mecab.parse(text).splitlines()

  # 解析結果を格納するリスト
  results = []

  for line in parsed_lines:
      # 'EOS' (End of Sentence) はスキップ
      if line == 'EOS':
          break

      # 表層形と素性情報をタブで分割
      surface, feature = line.split('\t')

      # 素性情報をカンマで分割
      features = feature.split(',')

      # 辞書（ハッシュマップ）形式で情報を整理
      morpheme = {
          'surface': surface,            # 表層形
          'pos': features[0],            # 品詞
          'pos_detail1': features[1],    # 品詞細分類1
          'pos_detail2': features[2],    # 品詞細分類2
          'pos_detail3': features[3],    # 品詞細分類3
          'conjugation_type': features[4],# 活用型
          'conjugation_form': features[5],# 活用形
          'base_form': features[6],      # 基本形（原型）
          'reading': features[7] if len(features) > 7 else '', # 読み
          'pronunciation': features[8] if len(features) > 8 else '' # 発音
      }
      results.append(morpheme)
  return results
```

```
json_result = mecab_json('昨日、柔らかくて固くて白くて高いバイトに行きました')
print(json_result)
for word in json_result:
  if (word['pos'] == '名詞' or word['pos'] == '形容詞'):
    print(word['pos'] + ":" + word['surface'])
```
