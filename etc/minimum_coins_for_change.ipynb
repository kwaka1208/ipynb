```python
# 硬貨の種類
coins = [500,10,50,5,100,1]

# 硬貨の種類を表示
print('硬貨の種類は', coins, 'です')

# 金額の大きい順に並べ替え
coins.sort(reverse=True)

# 硬貨の種類を表示
print('金額の大きい順に並べ替えると', coins, 'です')

# 買った金額
kingaku = 128

# 支払った金額
shiharai = 1000

# おつりを計算
otsuri = shiharai - kingaku #
print('おつりは',otsuri,'円です')

# 硬貨の枚数
coin_maisuu = 0

# コインの枚数を金額の大きいもの順に計算
for coin in coins:
  maisuu = otsuri // coin  # おつり
  print (coin, '円玉が', maisuu, '枚')
  otsuri = otsuri - maisuu * coin
  coin_maisuu = coin_maisuu + maisuu

# 硬貨の枚数の合計を表示
print('硬貨の枚数の合計は',coin_maisuu,'枚です')
```
