whileを使って硬貨を1枚1枚数える方式

```python
maisuu = 0
money = 12325
coins = [500,100,50,10,5,1]
for coin in coins:
  while money >= coin:
    money = money - coin
    maisuu = maisuu + 1
print(maisuu)
```

商を求める演算子と剰余（割り算のあまり）を求める演算子を使う方法

```python
goukei_maisuu = 0
money = 12325
coins = [500,100,50,10,5,1]
for coin in coins:
  maisuu = money // coin # 必要な硬貨の枚数を計算
  goukei_maisuu = goukei_maisuu + maisuu # 硬貨ごとに必要枚数を加算
  money = money % coin # 残額を計算
  print(f' {coin} 円が {maisuu} 枚')
print(goukei_maisuu)
```
