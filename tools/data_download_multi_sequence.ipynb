```python
import requests
from google.colab import drive
from PIL import Image
from io import BytesIO
import os

# Google Driveのマウントポイント
drive_root_dir ='/content/drive'
drive.mount(drive_root_dir)

# Google Driveの中のフォルダ名
save_folder = '/My Drive/okeiko/'

# ダウンロード元のURL
url = 'https://docs.crssrds.jp/assets/learn/target-data/'

# 保存先のフォルダの場所をマウントポイント + フォルダ名で作成。
target_folder = drive_root_dir + save_folder

# そのフォルダがあるかどうかチェック
if not os.path.exists(target_folder):
  # なければ作成する
  os.mkdir(target_folder)

for number in range(1,11):
  number = str(number)
  try :
    # URLにアクセス
    response = requests.get(url+"dog"+number+".jpg")
    print(url+"dog"+number+".jpg")
   # 結果のコードをチェック
    print(response.status_code)
    # 読み込んだデータを画像に変換
    img = Image.open(BytesIO(response.content))

    # 画像を保存先フォルダに保存
    img.save(target_folder + "dog"+number+".jpg")
    print('ダウンロード完了：',"dog"+number+".jpg")
  except Exception as e:
    # エラーが発生したらエラーコードを表示
    print('エラーが発生しました')
    print(e)
```
