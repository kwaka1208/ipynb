# チーム分けプログラム
クラスの中を指定した数のチーム数でチーム分けするプログラムです。

以下の手順で実行してください。

1. チーム分けする学年をmember変数で指定します（プログラム中にコメントで記載しています）
2. プログラムを実行します。
3. チーム数入力になるので、いくつに分けたいかを入力してください。割り切れない場合でも、適切にチーム分けを行います。
4. 出力にチームごとの名前が表示されます。

欠席者など除外したい人がいる場合は、配列の該当行をコメントアウトしてください。

```python
import random

member1 = [
"植山雄斗"
,"川上大心"
,"菊地寿弥"
,"坂口航"
,"田口慧都"
,"角田龍亮"
,"野村亮太"
,"馬場凜音"
,"松本伊織"
,"山口智也"
,"横内功士朗"
,"吉田京太郎"
]

member2 = [
"石田雄樹"
,"伊藤悠太"
,"上林鈴華"
# ,"大久保健斗"
,"金田航輝"
,"酒井奨眞"
,"阪井拓海"
,"嶋川知宜"
,"関場亮太郎"
,"民谷友菜"
# ,"津賀貴斗"
,"内藤颯"
# ,"中山大翔"
,"橋本元"
,"三嶋龍二"
# ,"南翔葵"
,"本谷太輝"
,"森藤千佳"
,"森脇寛斗"
,"山下龍一"
]

member3 = [
"大井一真"
,"北村壮"
,"田口暉継"
,"立道太樹"
,"中野黎音"
,"中森涼太"
,"細見謙太"
,"松浪圭吾"
]

#
# チーム分けする学年をここで指定します。
# member = member1 # 1年生の場合
# member = member2 # 2年生の場合
# member = member3 # 3年生の場合
#
member = member2

memberCount = len(member)
print("今日の出席者は" + str(memberCount)  + "人です。")
teamCount = int(input("何チームに分けますか？"))
mod = memberCount % teamCount
memberPerTeam = int((memberCount - mod) / teamCount)

teams = []
for i in range(teamCount):
  count = memberPerTeam
  if(mod > 0):
    count+= 1
  mod-=1
  single = []
  for j in range(count):
    pick = random.randint(1, len(member)) - 1
    single.append(member.pop(pick))
  teams.append(single)

for team in teams:
  print(team)
  





```
