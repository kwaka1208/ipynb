```python
import requests
from google.colab import drive
from PIL import Image
from io import BytesIO
import os

# Google Driveのマウントポイント
drive_root_dir ='/content/drive'
drive.mount(drive_root_dir)

# Google Driveの中のフォルダ名
save_folder = '/My Drive/QR/'

base_url = "https://chart.apis.google.com/chart?chs=300x300&cht=qr&chl="

# ダウンロード元のURL
target_url = 'https://dojocon2023.coderdojo.jp/expo/'

# 保存先のフォルダの場所をマウントポイント + フォルダ名で作成。
target_folder = drive_root_dir + save_folder

# そのフォルダがあるかどうかチェック
if not os.path.exists(target_folder):
  # なければ作成する
  os.mkdir(target_folder)

try :
  for i in range(1, 36):
    # URLにアクセス
    response = requests.get(base_url + target_url + str(i))
    filename = str(i) + ".png"
    # 結果のコードをチェック
    print(response.status_code)
    # 読み込んだデータを画像に変換
    img = Image.open(BytesIO(response.content))


    # 画像を保存先フォルダに保存
    img.save(target_folder + filename)
    print('ダウンロード完了：',filename)
except Exception as e:
  # エラーが発生したらエラーコードを表示
  print('エラーが発生しました')
  print(e)
```
