```python
#
# あるチームを複数の小さなチームに分けるためのプログラムです。
# 学校の授業などで、その時だけのチームや班を作るのに使えます。
# 最初に出席者のリストmemberから欠席者をコメントアウトして実行してください。
# 現在の出席者数を表示しますので、何チームに分けたいかチーム数を入力してください。
# メンバー数とチーム数から適当な人数に割り振ったチームを作ります。
#

import random

#
# 全メンバーのリストです。こちらを適宜書き換えてください。
# 欠席者がいる場合は削除せずにコメントアウト（行の先頭に#をつける）すると良いでしょう。
#
member = [
# 以下のメンバーは架空の人物たちです
  "今理沙"
  ,"有賀小絵"
  ,"大矢莉菜"
  ,"麻生優那"
  ,"真田満里奈"
  ,"堀内佐登子"
  ,"二瓶陽保"
  ,"三木嘉一"
  ,"宮内紳悟"
  ,"河井美和"
]

memberCount = len(member)
print("今日の出席者は" + str(memberCount)  + "人です。")
teamCount = int(input("何チームに分けますか？"))
mod = memberCount % teamCount
memberPerTeam = int((memberCount - mod) / teamCount)

teams = []
for i in range(teamCount):
  count = memberPerTeam
  if(mod > 0):
    count+= 1
  mod-=1
  single = []
  for j in range(count):
    pick = random.randint(1, len(member)) - 1
    single.append(member.pop(pick))
  teams.append(single)

for team in teams:
  print(team)
  





```
