```python
import requests
import json

#
# 書籍情報
# ISBN-10（10桁のISBNコード）を入力してください
#
# 検索URL : http://api.openbd.jp/v1/get?isbn=isbn&pretty

#
# コードを生成する本を指定
#
language = [
  [
     "4862807801"
    ,"4426126304"
    ,"4591168018"
    ,"4774187747"
    ,"4774183687"
    ,"4297116391"
    ,"4074447541"
    ,"4299014502"
  ]
]

#
# 書籍情報を出力
#
def put_a_book(isbn):
  try:
    api_url = "http://api.openbd.jp/v1/get?isbn="
    request_url = api_url + isbn
    jsondata = requests.get(request_url).json()
    title_text = jsondata[0]["summary"]["title"]
    #
    # Cover Image 出力
    #
    cover_url = jsondata[0]["summary"]["cover"]
    if len(cover_url) == 0:
      cover_url = "/assets/images/no_image.png"
    print(f'title: {title_text}')
    print(f'cover: {cover_url}')
  except:
    print(f"error:{isbn}")
  print(f'')

#
# メイン処理
#
if __name__ == "__main__":
  for books in language:
    print ('--------------------------------')
    for isbn in books:
      put_a_book(isbn)
```
