```python
# Pythonの例（戻り値のない関数）

def circle(hankei):
    print("半径", hankei, "cmの時")
    print("面積は", hankei * hankei * 3.14, "cm2")
    print("\n")

circle(5)


# Pythonの例（戻り値のある関数）

def double(number):
    return number * 2

print(double(5))
```
