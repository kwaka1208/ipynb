````python
# Webサイトにアクセスするためのライブラリを使います
import requests

# Webページの中のデータにアクセスできるようにするためのライブラリも使います
from bs4 import BeautifulSoup

# 今回データを取得するメニューページのURLです
url = 'https://lab.crssrds.jp/y/'

# メニューページにアクセスして、中身のデータを変数 html に入れます。
html = requests.get(url)             

# ライブラリ'BeautifulSoup'を使って中のデータに自由にアクセスできるようにします。
soup = BeautifulSoup(html.content, 'html.parser') 
````

````python
# newsの見出し部分を取り出す。
news_block = soup.find(id='tabpanelTopics1')
print(news_block)
````

````python
# 見出し部分から一覧を取り出す
topics = news_block.find_all('span', class_='fQMqQTGJTbIMxjQwZA2zk _1alzSpTqJzvSVUWqpx82d4')
for topic in topics:
  print(topic.text)
````

見出し部分からそのまま一覧を取り出すと余計なものが混ざっている。

webサイトのソースの中をさらにみていくと、972行目に'ul'タグがあり、この中に見出しが入っていることがわかる。

https://github.com/kwaka1208/lab/blob/main/y/index.html#L972

この'ul'の中だけをさらに取り出して、絞り込んでいく。

````python
# news見出し部分ブロックからさらに見出しの一覧だけを取り出す
topics_black = news_block.find('ul')
topics = topics_black.find_all('span', class_='fQMqQTGJTbIMxjQwZA2zk _1alzSpTqJzvSVUWqpx82d4')
#print(topics)

for topic in topics:
  print(topic.text)
````

webページの構造に注目すれば、
  ```
 class_='fQMqQTGJTbIMxjQwZA2zk_1alzSpTqJzvSVUWqpx82d4')
 ```
 を使わなくても構造をたどるだけでも見つけられる。

````python
# news見出し部分ブロックからさらに見出しの一覧だけを取り出す
topics_black = news_block.find('ul')


topics = topics_black.find_all('li')
for topic in topics:
  topic = topic.find('span')
  print(topic.text)
````
