## 基本編
## テキストを表示してみよう
さっそく、テキスト（文章）を表示させてみましょう。
表示には、printという命令を使います、テキストを表示させる場合には表示したいテキストを「'」（シングルクォーテーション）か「"」（ダブルクォーテーション）で囲まなければなりません。

では、プログラミング初心者の定番メッセージ「Hello, World」を表示してみましょう。

```
print("Hello, World")
```
「print」は英語で印刷の意味ですが、プログラムの命令では表示を意味しています。

"Hello, World"という言葉を表示せよという命令です。

上のプログラムを入力して、実行してみてください。

どうしても難しければ、上のプログラムをコピー＆ペーストしても構いませんが、できるだけ自分の手で入力してみましょう。


## 数字を表示してみよう
次に数字を表示します。

単に数字を表示するだけでは面白くないので、コンピューターに計算させて計算した結果を表示させてみます。

```
print(5 + 3)
```

このプログラムを入力して実行してみてください。

## 文字と数字の組み合わせて表示してみよう

次に文字と数字を組み合わせて表示します。

```
print("5 + 3 = ", 5+3)
```
左側の「5 + 3 =」は"で囲まれているのでテキストとして扱われます。

「,」の隣の 5+3 は数字ですので計算結果が表示されます。

では、このプログラムを入力して実行してみましょう。

## 四則演算をやってみよう
先ほどのプログラムでは、計算式と答えが表示されました。

では、残りの掛け算、割り算、引き算の式と答えを表示するプログラムを作ってみましょう。

足し算の記号は「+」、引き算の記号は「-」と普通の数学と同じですが、掛け算は「*」、割り算は「/」を使います。これは他の多くのプログラミング言語で共通で、コンピュータでは、掛け算は「*」、割り算は「/」で表されます。

|||
|:--:|:--:|
|足し算|+|
|引き算|-|
|掛け算|*|
|割り算|/|
