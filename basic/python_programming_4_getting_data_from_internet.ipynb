````python
# Webサイトにアクセスするためのライブラリを使います
import requests

# Webページの中のデータにアクセスできるようにするためのライブラリも使います
from bs4 import BeautifulSoup

# 今回データを取得するメニューページのURLです
url = 'https://lab.crssrds.jp/menu/'

# メニューページにアクセスして、中身のデータを変数 html に入れます。
html = requests.get(url)             

# ライブラリ'BeautifulSoup'を使って中のデータに自由にアクセスできるようにします。
soup = BeautifulSoup(html.content, 'html.parser') 
````

### 読み込んだWebページのデータを表示してみよう


````python
print(soup)
````

### お店の名前を取得しよう

![](https://lab.crssrds.jp/menu/images/01_shop_name.png)

お店の名前は'h1'というタグで書かれているので、h1を探してその内容を表示します。

````python
# h1の要素を探します
element = soup.find('h1')

# 見つけた要素を表示します
print(element) 
````

````python
# 要素の中から内容（テキスト）だけを取り出します
print(element.text)
````

### メニューの中身を全部取り出そう

メニューの内容は'li'というタグで書かれているので、liを全部探してその内容を表示します。

![](https://lab.crssrds.jp/menu/images/02_all_menu.png)

````python
# find_all は指定した要素をすべて探します。
elements = soup.find_all('li') 

# 見つけた要素（複数）を全部表示します
print (elements)
````

````python
# 取り出したすべての要素から内容を順番に表示します。
for element in elements:
  print(element.text)
````

### メニューの種類（フード、ドリンク、サイド）だけを取り出そう
先ほどのプログラムでは、メニューの種類と商品名が混ざっていたので、メニューの種類（フード、ドリンク、サイド）だけを取り出してみます。

````python
# メニューの種類は、li要素の中の"title"というclassがついているのでclassを指定して探します。
elements = soup.find_all('li', class_="title") # class_になっているので注意！

# elementsの中に3つの要素が入っていることを確認します。
print(elements)
````

````python
# 取り出した3つの要素から内容を表示します。
for element in elements:
  print(element.text)
````

### for の使い方
for は順番に並んだデータや数値の中からひとつずつデータを取り出すのに使われます。

```
elements = soup.find_all('li', class_="title")
```
を実行した結果、dataの中には以下の情報が入っています。

```
elements = ['フード', 'ドリンク', 'サイドメニュー']
```
これらの3つのデータをfor で順番に取り出していきます。

```
for element in elements: # elements の中から1つデータを取り出してelementに入れる
  print(element.text) # elementの内容を表示する
# elementsの中に他のデータがあれば、次のデータを取り出して繰り返す。
```
この処理をelementsの中のデータの数分だけ繰り返します。
