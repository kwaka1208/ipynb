```python
# Webサイトにアクセスするためのライブラリを使います
import requests

# Webページの中のデータにアクセスできるようにするためのライブラリも使います
from bs4 import BeautifulSoup

# 今回データを取得するメニューページのURLです
url = 'https://lab.crssrds.jp/y/'

# メニューページにアクセスして、中身のデータを変数 html に入れます。
html = requests.get(url)             

# ライブラリ'BeautifulSoup'を使って中のデータに自由にアクセスできるようにします。
soup = BeautifulSoup(html.content, 'html.parser') 
```

```python
# newsの見出し部分を取り出す。
news_block = soup.find(id='tabpanelTopics1')
print(news_block)
```

```python
# newsの見出し部分を取り出す。
titles = news_block.find_all('span', class_="fQMqQTGJTbIMxjQwZA2zk _1alzSpTqJzvSVUWqpx82d4")
for title in titles:
  print(title)
```

```python
# newsの見出し部分を取り出す。
for title in titles:
  print(title.text)
```
