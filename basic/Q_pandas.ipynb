# Pandas 演習: 名簿と平均点の計算

以下のデータを Pandas の DataFrame にして、2つの計算を行ってください。

1.  **数学の平均点**（列の平均）を求めてください。
2.  **鈴木さんの平均点**（行の平均）を求めてください。

| 名前 | 数学 | 英語 |
| :--- | :--- | :--- |
| 山本 | 85 | 92 |
| 鈴木 | 72 | 80 |
| 高橋 | 95 | 88 |

## ヒント1: データの作成とインデックスの設定
辞書形式から DataFrame を作る際、`index` を指定すると特定の行（人）を選びやすくなります。

```python
import pandas as pd

temp_data = {
    "項目1": [10, 20],
    "項目2": [30, 40]
}
# 練習用のDataFrame作成例
temp_df = pd.DataFrame(temp_data, index=["Aさん", "Bさん"])
display(temp_df)
```

## ヒント2: 列の計算（科目の平均など）
`df["列名"]` で特定の列を取り出し、`.mean()` でその平均を計算できます。

```python
# 「項目1」の平均を出す例
avg_col = temp_df["項目1"].mean()
print(f"項目1の平均: {avg_col}")
```

## ヒント3: 行の計算（個人の平均など）
特定の行を選ぶには `.loc["名前"]` を使います。その後に `.mean()` をつければ、その行の数値の平均が得られます。

```python
# 「Aさん」のデータの平均を出す例
avg_row = temp_df.loc["Aさん"].mean()
print(f"Aさんの平均: {avg_row}")
```
