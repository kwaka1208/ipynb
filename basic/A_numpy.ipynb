# NumPy 演習: 解答例と解説

2から18までの偶数を並べた 3x3 の行列を作成するプログラムです。
初心者の方にも分かりやすいよう、ステップバイステップで解説します。

## 方法1: `np.arange` の「ステップ」を使う方法

もっとも直接的な方法です。

1.  **ステップ1**: `np.arange(2, 20, 2)` で、2から始まり20未満まで2ずつ増える数値の並び（1次元配列）を作ります。
2.  **ステップ2**: `.reshape(3, 3)` で、その9個の数値を3行3列の形に並べ替えます。

```python
import numpy as np

# 1. 2から20未満まで、2飛ばしで数値を作る
arr = np.arange(2, 20, 2)
print("手順1（1次元配列）:")
print(arr)

# 2. 3行3列に形を整える
matrix1 = arr.reshape(3, 3)
print("\n手順2（3x3行列）:")
print(matrix1)
```

## 方法2: 「ブロードキャスト」を使う方法

NumPyの強力な機能である一括計算を利用する方法です。

1.  **ステップ1**: `np.arange(1, 10)` で、基本となる1から9の数値を作ります。
2.  **ステップ2**: `* 2` を計算します。これにより、すべての要素が一度に2倍されます（これがブロードキャストです）。
3.  **ステップ3**: 最後に `.reshape(3, 3)` で行列の形にします。

```python
# 1. 1〜9 の数値を作る
base = np.arange(1, 10)

# 2. 全体を2倍にする（ブロードキャスト）
doubled = base * 2

# 3. 3行3列にする
matrix2 = doubled.reshape(3, 3)

print("ブロードキャストを使った結果:")
print(matrix2)
```
