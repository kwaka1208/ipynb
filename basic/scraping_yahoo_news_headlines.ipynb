```python
# Webサイトにアクセスするためのライブラリを使います
import requests

# Webページの中のデータにアクセスできるようにするためのライブラリも使います
from bs4 import BeautifulSoup

# Yahoo! JAPANのトップページ
url = 'https://www.yahoo.co.jp/'

# メニューページにアクセスして、中身のデータを変数 html に入れます。
html = requests.get(url)
# print(html.text)

# ライブラリ'BeautifulSoup'を使って中のデータに自由にアクセスできるようにします。
soup = BeautifulSoup(html.content, 'html.parser')

# newsの見出し部分を取り出す。
news_block = soup.find(id='tabpanelTopics1')
# print(news_block)

# news見出し部分ブロックからさらに見出しの一覧だけを取り出す
topics_black = news_block.find('ul')
# print(topics_black)

topics = topics_black.find_all('li')
for topic in topics:
  print(topic.text)
```

## Amazonの検索結果からカテゴリーのIDを取得する

```python
import requests
from bs4 import BeautifulSoup

# 探索対象のURL
url = 'https://www.amazon.co.jp/s?k=%E3%81%88%E3%82%93%E3%81%B4%E3%81%A4&__mk_ja_JP=%E3%82%AB%E3%82%BF%E3%82%AB%E3%83%8A&crid=T798QLDFOLKG&sprefix=%E3%81%88%E3%82%93%E3%81%B4%E3%81%A4%2Caps%2C210&ref=nb_sb_noss_1'

# メニューページにアクセスして、中身のデータを変数 html に入れます。
html = requests.get(url)

# ライブラリ'BeautifulSoup'を使って中のデータに自由にアクセスできるようにします。
soup = BeautifulSoup(html.content, 'html.parser')
cats = soup.select("div#departments > ul > li")
for cat in cats:
  print(cat.get('id'))
```
