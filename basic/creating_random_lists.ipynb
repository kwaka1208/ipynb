## 課題
0 〜 n-1の数値をランダムに並べた配列を作るプログラム。ただし、配列内に同じ数字は入らないようにする。

例）n = 10の場合

OK:[4, 5, 0, 1, 8, 6, 3, 2, 9, 7]（0〜9の数字がひとつずつ入っている）

NG:[4, 5, 0, 1, 8, 1, 3, 2, 7, 7]（1と7がそれぞれ2つずある）

使用する言語は問いません。できた方は、n = 10000に設定してこの投稿のコメントでプログラムへのリンクを張るかデータを添付してください。つまり、10,000個の数字をランダムに並べた配列を作るのが最終目標。

## 実装例

```python
#
# 最大値の設定（各実装例の実行前にこのブロックを実行してください）
#
max_data = 10000
```

```python
# 実装例1
# 乱数を発生させて配列に追加していく。既に作成済みの数字が出たら乱数を作り直す
# 手あたり次第に乱数を発生させているので、かなり時間がかかります。
import time
import random

random_list = []
#
# 元データの作成
#
start_time = time.time()
for i in range(max_data):
  while True:
    sample = random.randrange(max_data)
    if(random_list.count(sample) == 0):
      # 配列にない数値だったら配列に追加それ以外なら乱数の発生からやり直し
      random_list.append(i)
      break

print(time.time() - start_time)
print(random_list)
```

```python
# 実装例2
# 順番に並んだデータを作って、ランダムに取り出して並べ替える
#
import time
import random

base_data = []
random_list = []

#
# 元データの作成
#
start_time = time.time()
for i in range(max_data):
  base_data.append(i)

#
# 元データからランダムにデータを取り出して、別の配列に追加
#
while len(base_data) != 0:
  index = random.randrange(len(base_data))
  random_list.append(base_data[index])
  # 一度使ったデータは重複を避けるために削除する
  del base_data[index] 

print(time.time() - start_time)
print(random_list)
```

```python
# 実装例3
# list関数で配列を作ってshuffleで混ぜる
#
import random
import time

start_time = time.time()
random_list = list(range(max_data))
random.shuffle(random_list)
print(time.time() - start_time)
print(random_list)
```

```python
# 実装例4
# random.sampleで指定範囲の乱数を重複無しで発生させて配列に格納する
#
import random
import time

start_time = time.time()
random_list = random.sample(range(max_data), max_data)

print(time.time() - start_time)
print(random_list)
```
