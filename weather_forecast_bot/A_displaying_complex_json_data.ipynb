## やってみよう
上のプログラムを参考に、すべての商品の情報（名前、在庫、値段）を表示するプログラムを作ってみてください。

ひとつずつ表示してもいいし、できる人は繰り返しを使ってみてもOKです。

```python
#
# forで表示する方法（１）
#
# こちらのプログラムでは、真ん中の数字の部分をfor in range(3)を使って0〜2まで変化させて表示します。
#
for i in range(3):
  print('商品名：',jsondata['fruit'][i]['name'], '在庫数：',jsondata['fruit'][i]['name'], '価格：',jsondata['fruit'][i]['price'], )

for i in range(3):
  print('商品名：',jsondata['vegetable'][i]['name'], '在庫数：',jsondata['vegetable'][i]['name'], '価格：',jsondata['vegetable'][i]['price'], )
  
```

```python
#
# forで表示する方法（２）
#
# こちらのプログラムでは、繰り返しの部分をひとつずつ取り出して表示しています。
# このやり方だと、中にいくつのデータがあるかを気にしなくてもよいので、データの個数が変わってもプログラムを変えなくても良いのが特長です。
#
for i in jsondata['fruit']:
  print('商品名：',i['name'], '在庫数：', i['name'], '価格：',i['price'], )

for i in jsondata['vegetable']:
  print('商品名：',i['name'], '在庫数：', i['name'], '価格：',i['price'], )
  
```

## なぜ「forで表示する方法（２）」では数字がないのか？

「forで表示する方法（２）」のiのところで数字がいらないのはなぜか？という質問をいただいたので、iの中身を表示するプログラムを作ってみました。

```python
count = 0
for i in jsondata['fruit']:
  print('---')
  print(count, '番目', end=':')
  print(i)
  count = count + 1

count = 0
for i in jsondata['vegetable']:
  print('---')
  print(count, '番目', end=':')
  print(i)
  count = count + 1
```

実行結果を見てみると、iを表示するたびに商品一つ分のJSONデータが表示されます。つまり、「forで表示する方法（２）」のプログラムでは、繰り返し部分のデータをひとつずつ取り出しているので数字で指定しなくてもいいのです。

これはリストで繰り返した場合と同じですね。

```python
#
# リストの中の分だけ繰り返す
#
month = ["睦月","如月","弥生","卯月","皐月","水無月","文月","葉月","長月","神無月","霜月","師走"]

for m in month:
  print(m)
```

## やってみよう２
それぞれの商品が全部売れた時の金額（売上金額）を計算するプログラムを作ってみよう。

売上金額は、在庫数 x 値段で計算できます。

```python
#
# ここでは各商品ごとの合計、果物の合計、野菜の合計、全部の合計を計算してみました。
#

fruit_total = 0
vegetable_total = 0

for i in jsondata['fruit']:
  amount = i['stock'] * i['price']
  fruit_total = fruit_total + amount

  print('商品名：',i['name'], '売上：', amount, '円' )
print('果物売上合計：',fruit_total, '円')

for i in jsondata['vegetable']:
  amount = i['stock'] * i['price']
  vegetable_total = vegetable_total + amount

  print('商品名：',i['name'], '売上：', amount, '円' )
print('野菜売上合計：',vegetable_total, '円')
print('----')
print('売上合計：',fruit_total + vegetable_total,'円')
```
