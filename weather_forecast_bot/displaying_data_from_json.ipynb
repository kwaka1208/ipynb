```python
# JSONデータを取り扱うモジュールです
import json

# 今回は、JSONデータをプログラムの中に用意します。
item = '{"name": "りんご", "stock": 50, "price": 300}'

# このデータを"json"モジュールの"loads"関数を使ってJSONデータとして読み込みます
jsondata = json.loads(item)

# 読み込んだJSONデータ全体を表示してみます。
print(jsondata)

#
name = jsondata['name'] # キーを使って商品名を取り出します
print(name)

# 同じように、在庫(stock)と価格(price)のデータを表示してみましょう。
```
