````python
import requests
import json

base_url = "https://api.openweathermap.org/data/2.5/weather" #APIの基本となる部分
parameter = "?zip={zipcode},{countrycode}&appid={apikey}&lang=ja" #APIに渡すパラメータの部分

# 郵便番号
zip = "100-0001"

# APIKEY 
key = ""  # ""の中に自分のAPIKEYを入れてください。

url = base_url + parameter
request_url = url.format(zipcode=zip, countrycode="jp",apikey=key)

# WebAPI実行用のURLを表示
print(request_url)

# WebAPIの実行結果を表示
data = requests.get(request_url)
print(data.text)
jsondata = data.json()
print(jsondata)

print("郵便番号{zip}の天気は{weather}です".format(zip=zip, weather=jsondata['weather'][0]['description']))
````
