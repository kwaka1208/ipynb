## このプログラムの問題点
このプログラムで利用しているOpenWeatherでは、郵便番号の情報を"000-0000"のように3桁の数字と4桁の数字を"-"で区切って与えなければなりません。また、全角は受け付けず半角としなければならないのです。

このプログラムではユーザーが入力した情報をそのままOpenWeatherに渡しているため、人間がこの通り入力しなければエラーとなってしまいます。

人間は間違うものなので、入力間違いに対してはシステム（アプリ）側で対応しなければなりません。人間が発生させてしまうエラーケースとして以下のものが考えられます。

### エラーとなるケース
1. 全角で入力された時  
日本語入力ONのまま入力を行った時に発生します。

郵便番号ではなく住所を入れなければならないと勘違いするかもしれません。
1. ７桁の数字以外が入力された時  
これは単純な入力ミスが原因のことが多いです。
1. 郵便番号のハイフン入れ忘れ  
ユーザーはハイフンが必要だということを知らないかもしれません。

こちらも単純な入力ミス、もしくは覚え間違いが原因のことが多い。

以下で、これらのエラーケースにそれぞれ対応していきます。

今回は、郵便番号のパターンをチェックしたいので、「3桁の数字-4桁の数字」のパターンでチェックします。これを正規表現で書くと以下のようになります。

[0-9]{3}-[0-9]{4}

これは、0-9の数字のどれかが3回と、"-"と、0-9の数字のどれかが4回のパターンを表しています。

Pythonで正規表現を使う場合は、"re"というモジュールを使います。reとは英語で正規表現を意味する"Regular Expression"の頭文字から名付けられています。

[re --- 正規表現操作](https://docs.python.org/ja/3/library/re.html)

この正規表現モジュールを使って正規表現による入力値チェックを作っていきます。

````python
# 正規表現モジュールを使うための宣言を行います。
import re

#
# 以下の3行は動作確認用に入れているものです。実際の動作時にはコメントアウト（先頭に#をつける）します。
#
# zip = "1234567" # この行の先頭の #を削除してみてください。チェック結果として"<re.Match object; span=(0, 8), match='1234567'>"が変数matchに入ります。
# zip = "123-4567" # この行の先頭の #を削除してみてください。チェック結果として"<re.Match object; span=(0, 8), match='123-4567'>"が変数matchに入ります。
# zip = "123-45678" # この行の先頭の #を削除してみてください。チェック結果として"None"が変数matchに入ります。
# zip = "abcdef"   # この行の先頭の #を削除して実行してみてください。チェック結果として"None"が変数matchに入ります。

# チェック用のパターンをzip_patternという変数に作成します。
zip_pattern = re.compile(r"^[0-9]{3}-?[0-9]{4}$")

# 作成したパターンで変数zipの郵便番号をチェックします。
match = zip_pattern.fullmatch(zip)

# チェック結果を表示します。
print(match)

# 一致しなかった場合は、エラーメッセージを表示します。
if match == None:
  print("入力された郵便番号が正しくありません。もう一度最初からお願いします")
````

これで、郵便番号の形式"000-0000"以外の入力が行われた場合のエラー処理ができました。

## 郵便番号のハイフン入れ忘れ
次に郵便番号のハイフン入れ忘れに対処していきましょう。

先ほどの正規表現を使って、ハイフンが入ったバターンとハイフンが入っていないパターンを表現すると以下の通りとなります。

```
[0-9]{3}-[0-9]{4} # ハイフンあり
[0-9]{3}[0-9]{4} # ハイフンなし
```
この両方のパターンに一致する場合で、数字の部分だけを取り出し、取り出した数字を"-"で繋ぐことで必ずハイフンありの郵便番号となるようにします。

正規表現では、「文字がある場合とない場合の両方」という条件を"?"で表すことができます。つまり

```
[0-9]{3}-?[0-9]{4}
```
このように、"-"後ろに?をつけると"-"がある場合とない場合の両方を見つけることができます。

```
123-4567
1234567
```
の両方に合致させられるのです。

さらに、パターンの一部分を()で囲むことで、囲まれた部分をひとつのグループとして取り扱い、グループごとに合致した部分を取り出せます。

例えば、3桁の数字の部分と4桁の数字の部分をそれぞれ取り出す場合には、以下のように()で囲みます。

```
([0-9]{3})-?([0-9]{4})
```

()で囲んだ部分を取り出すために、先ほどのfullmatch関数の代わりにsearch関数を使います。
fullmatch関数では文字列全体をチェックしますが、search関数は部分一致もチェックするため、1234-56789が入力された場合も"234-5678"の部分が一致してしまい正しい郵便番号と判断してしまいます。
このようになることを防ぐために、パターンに文字列の先頭と最後を表す^と$を加え以下の通りとします。

```
^([0-9]{3})-?([0-9]{4})$
```
このようにすることで、前後に数字がついていた場合は正しくない郵便番号として処理できます。

以上の内容をプログラムに反映していきます。

````python
# 正規表現モジュールを使うための宣言を行います。
import re

zip = "1234567"

# チェック用のパターンをzip_patternという変数に作成します。
# [0-9]{3}と[0-9]{4}の部分をそれぞれ()で囲みます。
zip_pattern = re.compile(r"^([0-9]{3})-?([0-9]{4})$")

# 作成したパターンで変数zipの郵便番号をチェックします。
# パターンに一致しない場合は"None"が入ります。
match = zip_pattern.search(zip)
# print(match)

# 一致しなかった場合は、エラーメッセージを表示します。
if match == None:
  print("入力された郵便番号が正しくありません。もう一度最初からお願いします")
else:
  # 一致した場合は、3桁の数字の部分と4桁の数字の部分を取り出して"-"でつなぎます。
  # group(1)がひとつめのグループ、group(2)が二つ目のグループを表します。
  zip = match.group(1) + "-" + match.group(2)

  # 最終的に処理された郵便番号が変数zipに入ります。
  print(zip)
````

このプログラムは、ユーザーが入力間違いしてもエラーで終了することはありません。
また、存在しない郵便番号を入力した場合もそのことを教えてくれます。

````python
!pip install zenhan

#
# OpenWeatherを使った天気予報bot
# https://openweathermap.org/
# 郵便番号を入力すると、その地域の天気予報を教えてくれます。
# ただし、入力エラーには未対応
#
# OpenWeatherが提供するAPIはこちら
# https://openweathermap.org/api

import requests
import json
import re
import zenhan

# APIの基本となる部分
base_url = "https://api.openweathermap.org/data/2.5/weather"

# パラメータ
parameter1 = "?zip="
parameter2 = ",jp&appid="
parameter3 = "&lang=ja"

# OpenWeatherを利用するためのAPIキーです。
# これをあとでURLに埋め込みます
key = ""

# 天気予報を表示したい地域の郵便場号をユーザーに入力してもらいます。
zip = input("天気を知りたい場所の郵便番号を入力してください")
# print(zip)
zip = zenhan.z2h(zip)
zip_pattern = re.compile(r"^([0-9]{3})-?([0-9]{4})$")

# 作成したパターンで変数zipの郵便番号をチェックします。
# パターンに一致しない場合は"None"が変数matchに入ります。
match = zip_pattern.search(zip)
# print(match)

# 一致しなかった場合は、エラーメッセージを表示します。
if match == None:
  print("入力された郵便番号が正しくありません。もう一度最初からお願いします")
else:
  # 一致した場合は、3桁の数字の部分と4桁の数字の部分を取り出して"-"でつなぎます。
  # group(1)がひとつめのグループ、group(2)が二つ目のグループを表します。
  zip = match.group(1) + "-" + match.group(2)

  # 郵便番号とAPIキーをURLの中に埋め込みます。
  request_url = base_url + parameter1 + zip + parameter2 + key + parameter3
  print(request_url)

  # 完成したURLでサービスにアクセスしてデータを取得します
  jsondata = requests.get(request_url).json()
  # print(jsondata)

  # 正しく天気予報の情報が取得できた時は、"cod"の値が200になります。
  if jsondata["cod"] == 200:
    # 実行結果を表示します。
    print("{place}の天気予報は以下の通りです".format(place=jsondata["name"]))
    print("気温：", jsondata["main"]["temp"])
    print("天気：", jsondata["weather"][0]["main"])
    print("天気詳細：", jsondata["weather"][0]["description"])
  else:
    print("入力された郵便番号は存在しませんでした。")

````
