```python
import requests
import json

base_url = "https://api.openweathermap.org/data/2.5/weather" #APIの基本となる部分
parameter = "?zip={zipcode},{countrycode}&appid={apikey}&lang=ja" #APIに渡すパラメータの部分

# 郵便番号
zip = "100-0001"

# APIKEY 
key = "<YOUR_API_KEY>" #APIキーを入れてください 

url = base_url + parameter
request_url = url.format(zipcode=zip, countrycode="jp",apikey=key)

# WebAPI実行用のURLを表示
print(request_url)

# WebAPIの実行結果を表示
print(requests.get(request_url).json())
```
